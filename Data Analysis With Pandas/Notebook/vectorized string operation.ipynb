## expand=True

```python
df["Name"].str.split(",").str.get(1).str.strip().str.split(" ",n=2,expand=True)
# convert into the dataframe 
```

```python
df[['title','firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
df.head()

df['title'].value_counts()
```

```python
# replace
df['title'] = df['title'].str.replace('Ms.','Miss.')
df['title'] = df['title'].str.replace('Mlle.','Miss.')
```

```python
df["title"].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['firstname'].str.endswith('A')]
# isdigit/isalpha...
df[df['firstname'].str.isdigit()]
```

```python
# lets find the name of the passenger whose first name is john 
df[df["surename"].str.contains("john",case=False)]
```

## regular expression 

```python
# find the last name of the person whose surename start with the vowel letter and end with the vowel letter
df[df["surename"].str.contains("^[aeiouAEIOU].+[aeiouAEIOU]$")]
```
